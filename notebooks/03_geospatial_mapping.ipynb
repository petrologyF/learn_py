## 0. 背景：DEMからの傾斜角算出
数値標高モデルから、土砂災害リスク等の評価に用いる傾斜角面を生成します。

## 課題：自動解析パイプラインの構築
学んだ手法を組み合わせて、一連の処理を完結させる関数を実装してください。

````
import rasterio
def run_module_pipeline():
    with rasterio.open('../data/raw_geospatial/dem.tif') as src:
        dem = src.read(1)
        res = src.res[0]
    dy, dx = np.gradient(dem, res)
    return np.rad2deg(np.arctan(np.sqrt(dx**2 + dy**2)))
````
