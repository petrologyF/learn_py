# #04 画像解析と結晶幾何学 (Scikit-Image / OpenCV)

## 0. 地球科学的背景: 薄片の組織解析
岩石薄片の顕微鏡写真から結晶のサイズ分布を求めることは、マグマの冷却速度や変成度を推定するための重要な手法です。

本実習では、画像の二値化（Thresholding）、ラベリング、そして各「結晶」の物理量（等価円直径など）の自動抽出を行います。

## 1. 画像処理のワークフロー (写経)
画像をデジタルデータ（行列）として扱い、特徴を抽出する手順を学びます。

```
from skimage import io, filters, measure

# 画像の読み込み
img = io.imread('../data/raw_images/thin_section.png', as_gray=True)

# 大津の二値化 (Otsu's Method)
thresh = filters.threshold_otsu(img)
binary = img > thresh

plt.imshow(binary, cmap='gray')
plt.title('Binary Image')
plt.show()
```

## 🎯 Challenge: 自動解析パイプラインの構築
これまでに学んだ手法を組み合わせ、1つの関数として実装してください。

```
from skimage import io, filters, measure

def run_module_pipeline():
    # --- STEP 1: 画像の読み込み ---
    # data/raw_images/thin_section.png をグレースケールで
    img = io.imread('../data/raw_images/thin_section.png', as_gray=True)
    
    # --- STEP 2: 二値化 ---
    # 結晶と石基を分離するための閾値を決定してください
    thresh = filters.threshold_otsu(img)
    binary = img > thresh
    
    # --- STEP 3: ラベリングと計測 ---
    # skimage.measure.label と regionprops を使用してください
    label_img = measure.label(binary)
    props = measure.regionprops(label_img)
    
    # --- STEP 4: 特徴量の抽出 ---
    # 各粒子の 'equivalent_diameter' (等価円直径) をリストに格納してください
    diameters = [p.equivalent_diameter for p in props]
    
    print(f"抽出された結晶数: {len(diameters)}")
    
    return diameters
```

## ✅ 検証と記録
実行してレポートをクリップボードにコピーします。

```
result = run_module_pipeline()
verify_and_log('M4', result)
```
